# ML Assignment 2 – IBM Watson Data Refinery

This Google Colab notebook loads and verifies the refined Retail Sales and Healthcare datasets prepared for Assignment 2.

**Datasets:**
- Retail Sales – refined CSV
- Healthcare – refined CSV


```python
import pandas as pd
from google.colab import files

print('Upload the two refined CSV files:')
uploaded = files.upload()
print('\nUploaded files:', list(uploaded.keys()))
```

```python
import os

retail_file = 'Retail_Sales_Refined.csv'
health_file = 'Healthcare_Refined.csv'

if not os.path.exists(retail_file) or not os.path.exists(health_file):
    raise FileNotFoundError(
        'Please upload both Retail_Sales_Refined.csv and Healthcare_Refined.csv.'
    )

retail = pd.read_csv(retail_file)
healthcare = pd.read_csv(health_file)

print('Retail Sales shape:', retail.shape)
print('Healthcare shape:', healthcare.shape)
```

## Q1 – Retail Sales Data Verification

```python
print('Retail Sales – first 10 records')
display(retail.head(10))

print('\nData types:')
display(retail.dtypes.to_frame('Data Type'))

print('\nMissing values:')
display(retail.isnull().sum().to_frame('Missing Values'))
```

```python
# Verify duplicate records and standardized order status
print('Exact duplicate rows:', retail.duplicated().sum())
print('Duplicate Order_ID values:', retail['Order_ID'].duplicated().sum())
print('Duplicate Customer_ID values:', retail['Customer_ID'].duplicated().sum())

print('\nOrder_Status values and counts:')
display(retail['Order_Status'].value_counts())

print('\nUnique date formats/values after refinement (first 10):')
display(retail['Order_Date'].head(10))
```

```python
# Verify the required retail fields are preserved
required_retail_columns = [
    'Order_ID', 'Customer_ID', 'Customer_Name', 'City',
    'Order_Date', 'Product', 'Quantity', 'Unit_Price', 'Order_Status'
]
print('Required columns preserved:', set(required_retail_columns).issubset(retail.columns))
print('Order date range:', retail['Order_Date'].min(), 'to', retail['Order_Date'].max())
```

## Q2 – Healthcare Patient Data Verification

```python
print('Healthcare – first 10 records')
display(healthcare.head(10))

print('\nData types:')
display(healthcare.dtypes.to_frame('Data Type'))

print('\nMissing values:')
display(healthcare.isnull().sum().to_frame('Missing Values'))
```

```python
# Verify Age handling and removal of TemporaryNotes
print('Missing Age values:', healthcare['Age'].isnull().sum())
print('TemporaryNotes column present:', 'TemporaryNotes' in healthcare.columns)

numeric_columns = ['Age', 'Glucose', 'Blood_Pressure']
print('\nNumeric columns:')
for col in numeric_columns:
    print(f'{col}: {healthcare[col].dtype}')

print('\nMissing values in numeric fields:')
display(healthcare[numeric_columns].isnull().sum().to_frame('Missing Values'))
```

```python
# Final assignment summary
summary = pd.DataFrame({
    'Dataset': ['Retail Sales', 'Healthcare'],
    'Rows After': [len(retail), len(healthcare)],
    'Columns After': [len(retail.columns), len(healthcare.columns)],
    'Duplicate Rows': [retail.duplicated().sum(), healthcare.duplicated().sum()],
    'Key Check': [
        'Dates/status standardized',
        'Age handled; TemporaryNotes removed'
    ]
})
display(summary)

print('Notebook verification completed successfully.')
```

### Expected refined results
- Retail Sales: **30 rows × 9 columns**, no duplicate rows, dates standardized to `YYYY-MM-DD`, and status values standardized to `Completed`, `Pending`, and `Cancelled`.
- Healthcare: **30 rows × 8 columns**, missing `Age` handled, `TemporaryNotes` removed, and `Age`, `Glucose`, and `Blood_Pressure` retained as numeric fields.
